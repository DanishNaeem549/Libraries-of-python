```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Graph")
plt.show()
```

```python
# customizing the line graph
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y,color='red',linestyle='--',marker='o',markersize=10,markerfacecolor='blue',markeredgewidth=2,markeredgecolor='black')
```

```python
x = [1,2,3,4,5]
y1 = [2,4,6,8,10]
y2 = [1,3,5,7,9]

plt.subplot(1,2,1)
plt.plot(x,y1,label='Line 1')

plt.subplot(1,2,1)
plt.plot(y1,x,label='Line 1')

plt.subplot(1,2,2)
plt.plot(y2,x,label='Line 1')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Graph 1")
plt.legend()

plt.subplot(1,2,2)
plt.plot(x,y2,label='Line 2')
```

```python
import pandas as pd
```

```python
df =pd.read_csv("practice.csv")
```

```python
df.info()
```

```python
sales_data = df.groupby('Product')['Unit_Price'].sum()
```

```python
sales_data.plot(kind='bar',color='green')
```

```python
df['Unit_Price'].plot(kind='hist',bins=10,color='orange')
```

```python
labels = ['A','B','C','D','E']
sizes = [15,30,45,10,5]
explode = (0,0.1,0,0,0)  # only "explode" the 2nd slice (i.e. 'B')
plt.pie(sizes, labels=labels, autopct='%1.1f%%', explode=explode)
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
labels = df['Product'].unique()
values = df.groupby('Product')['Unit_Price'].sum()

df1 =plt.pie(values, labels=labels, autopct='%1.1f%%')
plt.axis('equal')
plt.show()
```
