```python
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.datasets import cifar10
```

```python
(x_train, y_train), (x_test, y_test) = cifar10.load_data()
```

```python
x_train = x_train/255.0
x_test = x_test/255.0
```

```python
x_train.shape
```

```python
plt.imshow(x_train[0])
```

```python
model = tf.keras.models.Sequential()
model.add(tf.keras.layers.conv2D(filters=32, kernel_size=3, activation='relu',padding = 'same', input_size=(32,32,3)))
```

```python
model.add(tf.keras.layers.conv2D(filters=32, kernal_size=3,padding = 'same', activation= 'relu'))
model.add(tf.keras.layers.MaxPool(pool_size = 2,stride = 2, padding= 'valid'))
```

```python
model.add(tf.keras.layers.conv2D(filters=64, kernal_size=3,padding = 'same', activation= 'relu'))
```

```python
model.add(tf.keras.layers.conv2D(filters=64, kernal_size=3,padding = 'same', activation= 'relu'))
model.add(tf.keras.layes.Maxpool(pool_size=2, stride= 2, padding= 'valid'))
```

```python
model.add(tf.keras.layers.Flatten())
```

```python
model.add(tf.keras.layers.Dense(units=128, activation= 'relu'))
```

```python
model.add(tf.keras.layers.Dense(units=10, activation= 'softmax'))
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['sparse_categorical_accuracy'])
```

```python
model.fit(x_train, y_train, epochs=5)
```
